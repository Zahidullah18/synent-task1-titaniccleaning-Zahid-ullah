## 3. Handling Missing Values

Missing values can affect data analysis and may cause problems in later processing. Therefore, missing values are identified and analyzed before deciding how each affected column should be handled.


```python
df.isnull().sum()
```

### Missing Value Percentage

The percentage of missing values in each column is calculated to understand the extent of missing data.


```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_percentage.sort_values(ascending=False)
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

### Missing Value Treatment

The following approach is used to handle missing values:

* `Age`: Missing values are replaced using the median.
* `Embarked`: Missing values are replaced using the mode.
* `Cabin`: The column is removed because it contains a large proportion of missing values.


```python
df["Age"] = df["Age"].fillna(df["Age"].median())

df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

df.drop(columns=["Cabin"], inplace=True)
```

### Verification

After handling the missing values, the dataset is checked again to confirm that no missing values remain in the retained columns.


```python
df.isnull().sum()
```

## 4. Removing Duplicate Records

Duplicate records can cause the same observation to be counted more than once. Therefore, duplicate rows are identified and removed from the dataset.


```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

## 5. Data Type Conversion

The data types of the columns are reviewed and converted where appropriate to ensure that numerical and categorical variables use suitable data types.


```python
df.dtypes
```

```python
df["PassengerId"] = df["PassengerId"].astype("int64")
df["Survived"] = df["Survived"].astype("int64")
df["Pclass"] = df["Pclass"].astype("int64")
df["SibSp"] = df["SibSp"].astype("int64")
df["Parch"] = df["Parch"].astype("int64")

df["Age"] = df["Age"].astype("float64")
df["Fare"] = df["Fare"].astype("float64")
```

```python
df["Sex"] = df["Sex"].astype("category")
df["Embarked"] = df["Embarked"].astype("category")
```

```python
df.dtypes
```

## 6. Renaming Columns

The column names are standardized using lowercase letters and underscores to improve readability and maintain a consistent naming convention.


```python
df.rename(columns={
    "PassengerId": "passenger_id",
    "Survived": "survived",
    "Pclass": "passenger_class",
    "Name": "name",
    "Sex": "sex",
    "Age": "age",
    "SibSp": "siblings_spouses",
    "Parch": "parents_children",
    "Ticket": "ticket",
    "Fare": "fare",
    "Embarked": "embarked"
}, inplace=True)
```
